```python
from PIL import Image
import os
import glob

def batch_round_coordinates(input_folder, output_folder, decimal_places=5):
    """
    Process all JPG files and round coordinates to specified decimal places
    """
    os.makedirs(output_folder, exist_ok=True)
    
    jpg_files = glob.glob(os.path.join(input_folder, "*.jpg"))
    
    for jpg_file in jpg_files:
        try:
            filename = os.path.basename(jpg_file)
            name_without_ext = os.path.splitext(filename)[0]
            parts = name_without_ext.split('_')
            
            if len(parts) >= 4:
                # Round coordinates
                coord1 = round(float(parts[2]), decimal_places)
                coord2 = round(float(parts[3]), decimal_places)
                
                new_filename = f"{coord1}_{coord2}.png"
                output_path = os.path.join(output_folder, new_filename)
                
                # Convert and save
                img = Image.open(jpg_file)
                img.save(output_path, 'PNG')
                
                print(f"✓ {filename}")
                print(f"  → {new_filename}")
            else:
                print(f"✗ Skipped: {filename} (wrong format)")
                
        except Exception as e:
            print(f"✗ Error with {filename}: {e}")

# Usage
batch_round_coordinates("images_test", "images_png", decimal_places=)
```
