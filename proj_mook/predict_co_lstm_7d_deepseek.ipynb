```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
from scipy import fftpack
import warnings
warnings.filterwarnings('ignore')

# Set style for plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Load and preprocess data
df = pd.read_csv('CO_hex30_urban.csv')
df['date'] = pd.to_datetime(df['date'])
df.sort_values(['id', 'date'], inplace=True)

print("Dataset Information:")
print(f"Date range: {df['date'].min()} to {df['date'].max()}")
print(f"Number of sensors: {df['id'].nunique()}")
print(f"Total records: {len(df)}")

# Handle missing values using sensor-specific interpolation
# Fix for the time-based interpolation error
def interpolate_sensor_data(group):
    # Create a temporary dataframe with datetime index for interpolation
    temp_df = group.set_index('date').copy()
    temp_df['CO'] = temp_df['CO'].interpolate(method='time', limit_direction='both')
    temp_df['CO'] = temp_df['CO'].fillna(method='ffill').fillna(method='bfill')
    return temp_df['CO'].values

df['CO'] = df.groupby('id').apply(interpolate_sensor_data).explode().values

# Check if any nulls remain
print(f"Remaining null values: {df['CO'].isnull().sum()}")

# Aggregate daily average CO
daily_avg = df.groupby('date')['CO'].agg(['mean', 'std', 'min', 'max']).reset_index()
daily_avg.rename(columns={'mean': 'CO'}, inplace=True)

# Display descriptive statistics
print("\nDescriptive Statistics for Daily CO Levels:")
print(daily_avg['CO'].describe())

# 1. Histogram of daily mean CO values
plt.figure(figsize=(10, 6))
plt.hist(daily_avg['CO'], bins=30, edgecolor='black', alpha=0.7)
plt.title('Distribution of Daily Mean CO Values')
plt.xlabel('CO Concentration')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
# plt.savefig('co_histogram.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# 2. Time series with Fourier baseline
# Create time index
t = np.arange(len(daily_avg))

# Fit Fourier series (using FFT)
fft = fftpack.fft(daily_avg['CO'].values)
frequencies = fftpack.fftfreq(len(daily_avg))
power = np.abs(fft)

# Filter out high frequency noise
fft_filtered = fft.copy()
fft_filtered[np.abs(frequencies) > 0.1] = 0  # Adjust this threshold as needed

# Reconstruct signal
co_smooth = np.real(fftpack.ifft(fft_filtered))

plt.figure(figsize=(12, 6))
plt.plot(daily_avg['date'], daily_avg['CO'], label='Actual CO', alpha=0.7)
plt.plot(daily_avg['date'], co_smooth, label='Fourier Baseline', linewidth=2, color='red')
plt.title('CO Time Series with Fourier Baseline')
plt.xlabel('Date')
plt.ylabel('CO Concentration')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)
# plt.savefig('co_time_series_fourier.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# Create time series sequences for LSTM
def create_sequences(data, n_steps):
    X, y = [], []
    for i in range(len(data) - n_steps):
        seq = data[i:i+n_steps]
        X.append(seq)
        y.append(data[i+n_steps])
    return np.array(X), np.array(y)

n_steps = 14  # Use 14 days of history
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(daily_avg[['CO']])

X, y = create_sequences(scaled_data.flatten(), n_steps)

# Reshape X to 3D for LSTM [samples, timesteps, features]
X = X.reshape((X.shape[0], X.shape[1], 1))

# Train-test split (80-20)
split_idx = int(0.8 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

print(f"\nTraining samples: {X_train.shape[0]}")
print(f"Testing samples: {X_test.shape[0]}")

# Build enhanced LSTM model with bidirectional layers
model = Sequential([
    Bidirectional(LSTM(100, activation='relu', return_sequences=True), input_shape=(n_steps, 1)),
    Dropout(0.3),
    Bidirectional(LSTM(50, activation='relu')),
    Dropout(0.3),
    Dense(25, activation='relu'),
    Dense(1)
])

# Custom optimizer with learning rate scheduling
optimizer = Adam(learning_rate=0.001)
model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])

# Callbacks
early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=0.0001)

# Train model
print("\nTraining model...")
history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=16,
    validation_data=(X_test, y_test),
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# 4. Learning curve
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Training MAE')
plt.plot(history.history['val_mae'], label='Validation MAE')
plt.title('Model MAE')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
# plt.savefig('learning_curve.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# Evaluate model
y_pred = model.predict(X_test)

# Inverse transform to original scale
y_test_inv = scaler.inverse_transform(y_test.reshape(-1, 1))
y_pred_inv = scaler.inverse_transform(y_pred)

# Calculate metrics
rmse = np.sqrt(mean_squared_error(y_test_inv, y_pred_inv))
mae = mean_absolute_error(y_test_inv, y_pred_inv)
r2 = r2_score(y_test_inv, y_pred_inv)

print("\nModel Performance Metrics:")
print(f"RMSE: {rmse:.6f}")
print(f"MAE: {mae:.6f}")
print(f"R²: {r2:.6f}")

# 5. Actual vs Predicted
plt.figure(figsize=(10, 6))
plt.scatter(y_test_inv, y_pred_inv, alpha=0.6)
plt.plot([y_test_inv.min(), y_test_inv.max()], [y_test_inv.min(), y_test_inv.max()], 'r--', lw=2)
plt.xlabel('Actual CO Values')
plt.ylabel('Predicted CO Values')
plt.title(f'Actual vs Predicted CO Values (R² = {r2:.3f})')
plt.grid(True, alpha=0.3)
# plt.savefig('actual_vs_predicted.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# Forecast next 7 days
last_sequence = scaled_data[-n_steps:].flatten()
forecast = []

for _ in range(7):
    # Reshape for prediction
    seq = last_sequence.reshape(1, n_steps, 1)
    pred = model.predict(seq, verbose=0)
    
    # Store prediction
    forecast.append(pred[0, 0])
    
    # Update sequence for next prediction
    last_sequence = np.roll(last_sequence, -1)
    last_sequence[-1] = pred[0, 0]

# Inverse transform forecast
forecast = scaler.inverse_transform(np.array(forecast).reshape(-1, 1))

# Generate dates for forecast
last_date = daily_avg['date'].iloc[-1]
forecast_dates = [last_date + pd.Timedelta(days=i) for i in range(1, 8)]

# 3. 7-day forecast plot
plt.figure(figsize=(14, 7))

# Historical data
plt.plot(daily_avg['date'], daily_avg['CO'], label='Historical CO', linewidth=2)

# Fourier baseline
plt.plot(daily_avg['date'], co_smooth, label='Fourier Baseline', linewidth=2, color='green')

# Forecast
plt.plot(forecast_dates, forecast, label='7-Day Forecast', linewidth=2, color='red', marker='o')

plt.title('CO Concentration: Historical Data, Fourier Baseline, and 7-Day Forecast')
plt.xlabel('Date')
plt.ylabel('CO Concentration')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)
plt.tight_layout()
# plt.savefig('7_day_forecast.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# Display forecast values
print("\n7-Day Forecast:")
forecast_df = pd.DataFrame({
    'Date': forecast_dates,
    'CO_Forecast': forecast.flatten()
})
print(forecast_df.to_string(index=False))

# Print model summary
print("\nModel Architecture:")
model.summary()

# Create a comprehensive results summary
results_text = f"""
CO PREDICTION MODEL RESULTS
===========================

Dataset Information:
- Date range: {df['date'].min()} to {df['date'].max()}
- Number of sensors: {df['id'].nunique()}
- Total records: {len(df)}
- Remaining null values after processing: {df['CO'].isnull().sum()}

Descriptive Statistics for Daily CO Levels:
{daily_avg['CO'].describe().to_string()}

Model Performance Metrics:
- RMSE: {rmse:.6f}
- MAE: {mae:.6f}
- R²: {r2:.6f}

7-Day Forecast:
{forecast_df.to_string(index=False)}

Model Architecture:
{model.summary()}
"""

# Save results to file
with open('model_results.txt', 'w') as f:
    f.write(results_text)

print("\nAll visualizations and results have been saved to files:")
print("- co_histogram.png")
print("- co_time_series_fourier.png")
print("- learning_curve.png")
print("- actual_vs_predicted.png")
print("- 7_day_forecast.png")
print("- model_results.txt")
```
