```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
from scipy import fftpack
import warnings
warnings.filterwarnings('ignore')

# ตั้งค่าสไตล์สำหรับกราฟ
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# โหลดและเตรียมข้อมูล
df = pd.read_csv('CO_hex30_urban.csv')
df['date'] = pd.to_datetime(df['date'])
df.sort_values(['id', 'date'], inplace=True)

print("ข้อมูลของชุดข้อมูล:")
print(f"ช่วงวันที่: {df['date'].min()} ถึง {df['date'].max()}")
print(f"จำนวนเซ็นเซอร์: {df['id'].nunique()}")
print(f"จำนวนข้อมูลทั้งหมด: {len(df)}")

# จัดการค่าที่หายไปโดยใช้การประมาณค่าแบบเฉพาะเจาะจงสำหรับแต่ละเซ็นเซอร์
# แก้ไขข้อผิดพลาดการประมาณค่าแบบใช้เวลา
def interpolate_sensor_data(group):
    # สร้าง DataFrame ชั่วคราวที่มีดัชนี datetime สำหรับการประมาณค่า
    temp_df = group.set_index('date').copy()
    temp_df['CO'] = temp_df['CO'].interpolate(method='time', limit_direction='both')
    temp_df['CO'] = temp_df['CO'].fillna(method='ffill').fillna(method='bfill')
    return temp_df['CO'].values

df['CO'] = df.groupby('id').apply(interpolate_sensor_data).explode().values

# ตรวจสอบว่ายังมีค่า null เหลืออยู่หรือไม่
print(f"ค่าที่หายไปที่เหลืออยู่: {df['CO'].isnull().sum()}")

# รวมค่า CO เฉลี่ยรายวัน
daily_avg = df.groupby('date')['CO'].agg(['mean', 'std', 'min', 'max']).reset_index()
daily_avg.rename(columns={'mean': 'CO'}, inplace=True)

# แสดงสถิติเชิงพรรณนา
print("\nสถิติเชิงพรรณนาสำหรับระดับ CO รายวัน:")
print(daily_avg['CO'].describe())

plt.style.use('seaborn-v0_8')
# 1. Histogram ของค่า CO เฉลี่ยรายวัน
plt.figure(figsize=(12, 6))
plt.hist(daily_avg['CO'], bins=30, edgecolor='black', alpha=0.7)
plt.title('Distribution of Daily Mean CO Values')
plt.xlabel('CO Concentration')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
# plt.savefig('co_histogram.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# 2. อนุกรมเวลาพร้อมเส้นฐานฟูเรียร์
# สร้างดัชนีเวลา
t = np.arange(len(daily_avg))

# ปรับอนุกรมฟูเรียร์ (ใช้ FFT)
fft = fftpack.fft(daily_avg['CO'].values)
frequencies = fftpack.fftfreq(len(daily_avg))
power = np.abs(fft)

# กรองสัญญาณรบกวนความถี่สูง
fft_filtered = fft.copy()
fft_filtered[np.abs(frequencies) > 0.1] = 0  # ปรับเกณฑ์นี้ตามต้องการ

# สร้างสัญญาณใหม่
co_smooth = np.real(fftpack.ifft(fft_filtered))

plt.figure(figsize=(12, 6))
plt.plot(daily_avg['date'], daily_avg['CO'], label='Actual CO', alpha=0.7)
plt.plot(daily_avg['date'], co_smooth, label='Fourier Baseline', linewidth=2, color='red')
plt.title('CO Time Series with Fourier Baseline')
plt.xlabel('Date')
plt.ylabel('CO Concentration')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)
# plt.savefig('co_time_series_fourier.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# สร้างลำดับอนุกรมเวลาสำหรับ LSTM
def create_sequences(data, n_steps):
    X, y = [], []
    for i in range(len(data) - n_steps):
        seq = data[i:i+n_steps]
        X.append(seq)
        y.append(data[i+n_steps])
    return np.array(X), np.array(y)

n_steps = 14  # ใช้ประวัติ 14 วัน
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(daily_avg[['CO']])

X, y = create_sequences(scaled_data.flatten(), n_steps)

# ปรับรูปร่าง X เป็น 3D สำหรับ LSTM [samples, timesteps, features]
X = X.reshape((X.shape[0], X.shape[1], 1))

# แบ่งข้อมูลฝึกอบรมและทดสอบ (80-20)
split_idx = int(0.8 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

print(f"\nตัวอย่างการฝึกอบรม: {X_train.shape[0]}")
print(f"ตัวอย่างการทดสอบ: {X_test.shape[0]}")

# สร้างโมเดล LSTM ที่ปรับปรุงแล้วด้วยชั้น bidirectional
model = Sequential([
    Bidirectional(LSTM(100, activation='relu', return_sequences=True), input_shape=(n_steps, 1)),
    Dropout(0.3),
    Bidirectional(LSTM(50, activation='relu')),
    Dropout(0.3),
    Dense(25, activation='relu'),
    Dense(1)
])

# กำหนด optimizer พร้อมการกำหนดอัตราการเรียนรู้
optimizer = Adam(learning_rate=0.001)
model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])

# Callbacks
early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=0.0001)

# ฝึกโมเดล
print("\nกำลังฝึกโมเดล...")
history = model.fit(
    X_train, y_train,
    epochs=100,
    batch_size=16,
    validation_data=(X_test, y_test),
    callbacks=[early_stop, reduce_lr],
    shuffle=True,
    verbose=0
)

# 4. Learning curve
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Training MAE')
plt.plot(history.history['val_mae'], label='Validation MAE')
plt.title('Model MAE')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
# plt.savefig('learning_curve.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# ประเมินโมเดล
y_pred = model.predict(X_test)

# แปลงกลับสู่มาตราส่วนเดิม
y_test_inv = scaler.inverse_transform(y_test.reshape(-1, 1))
y_pred_inv = scaler.inverse_transform(y_pred)

# คำนวณเมตริก
rmse = np.sqrt(mean_squared_error(y_test_inv, y_pred_inv))
mae = mean_absolute_error(y_test_inv, y_pred_inv)
r2 = r2_score(y_test_inv, y_pred_inv)

print("\nเมตริกประสิทธิภาพของโมเดล:")
print(f"RMSE: {rmse:.6f}")
print(f"MAE: {mae:.6f}")
print(f"R²: {r2:.6f}")

# 5. Actual vs Predicted
plt.figure(figsize=(12, 6))
plt.scatter(y_test_inv, y_pred_inv, alpha=0.6)
plt.plot([y_test_inv.min(), y_test_inv.max()], [y_test_inv.min(), y_test_inv.max()], 'r--', lw=2)
plt.xlabel('Actual CO Values')
plt.ylabel('Predicted CO Values')
plt.title(f'Actual vs Predicted CO Values (R² = {r2:.3f})')
plt.grid(True, alpha=0.3)
# plt.savefig('actual_vs_predicted.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# พยากรณ์ 7 วันถัดไป
last_sequence = scaled_data[-n_steps:].flatten()
forecast = []

for _ in range(7):
    # ปรับรูปร่างสำหรับการพยากรณ์
    seq = last_sequence.reshape(1, n_steps, 1)
    pred = model.predict(seq, verbose=0)
    
    # บันทึกการพยากรณ์
    forecast.append(pred[0, 0])
    
    # อัปเดตลำดับสำหรับการพยากรณ์ครั้งต่อไป
    last_sequence = np.roll(last_sequence, -1)
    last_sequence[-1] = pred[0, 0]

# แปลงกลับสู่มาตราส่วนเดิม
forecast = scaler.inverse_transform(np.array(forecast).reshape(-1, 1))

# สร้างวันที่สำหรับการพยากรณ์
last_date = daily_avg['date'].iloc[-1]
forecast_dates = [last_date + pd.Timedelta(days=i) for i in range(1, 8)]

# 3. พล็อตการพยากรณ์ 7 วัน
plt.figure(figsize=(12, 6))

# ข้อมูลประวัติ
plt.plot(daily_avg['date'], daily_avg['CO'], label='Historical CO', linewidth=2)

# เส้นฐานฟูเรียร์
plt.plot(daily_avg['date'], co_smooth, label='Fourier Baseline', linewidth=2, color='green')

# การพยากรณ์
plt.plot(forecast_dates, forecast, label='7-Day Forecast', linewidth=2, color='red', marker='o')

plt.title('CO Concentration: Historical Data, Fourier Baseline, and 7-Day Forecast')
plt.xlabel('Date')
plt.ylabel('CO Concentration')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)
plt.tight_layout()
# plt.savefig('7_day_forecast.png', dpi=300, bbox_inches='tight')
# plt.close()
plt.show()

# แสดงค่าการพยากรณ์
print("\nการพยากรณ์ 7 วัน:")
forecast_df = pd.DataFrame({
    'Date': forecast_dates,
    'CO_Forecast': forecast.flatten()
})
print(forecast_df.to_string(index=False))

# พิมพ์สรุปโมเดล
print("\nโครงสร้างโมเดล:")
model.summary()

# สร้างสรุปผลลัพธ์แบบครอบคลุม
results_text = f"""
CO PREDICTION MODEL RESULTS
===========================

Dataset Information:
- Date range: {df['date'].min()} to {df['date'].max()}
- Number of sensors: {df['id'].nunique()}
- Total records: {len(df)}
- Remaining null values after processing: {df['CO'].isnull().sum()}

Descriptive Statistics for Daily CO Levels:
{daily_avg['CO'].describe().to_string()}

Model Performance Metrics:
- RMSE: {rmse:.6f}
- MAE: {mae:.6f}
- R²: {r2:.6f}

7-Day Forecast:
{forecast_df.to_string(index=False)}

Model Architecture:
{model.summary()}
"""

# บันทึกผลลัพธ์ไปยังไฟล์
with open('model_results.txt', 'w') as f:
    f.write(results_text)

print("\nการแสดงภาพและผลลัพธ์ทั้งหมดได้ถูกบันทึกไปยังไฟล์:")
print("- co_histogram.png")
print("- co_time_series_fourier.png")
print("- learning_curve.png")
print("- actual_vs_predicted.png")
print("- 7_day_forecast.png")
print("- model_results.txt")

# คำอธิบายขั้นตอนการทำงานในภาษาไทย
procedure_description = """
คำอธิบายขั้นตอนการทำงาน:

1. การเตรียมข้อมูล:
   - โหลดข้อมูลจากไฟล์ CSV และแปลงคอลัมน์วันที่เป็นรูปแบบ datetime
   - เรียงลำดับข้อมูลตาม ID เซ็นเซอร์และวันที่
   - จัดการค่าที่หายไปโดยใช้การประมาณค่าแบบเวลา (time interpolation) สำหรับแต่ละเซ็นเซอร์

2. การวิเคราะห์ข้อมูลเบื้องต้น:
   - คำนวณค่าเฉลี่ยรายวันของระดับ CO
   - สร้างฮิสโตแกรมเพื่อแสดงการกระจายของค่า CO รายวัน
   - ใช้การแปลงฟูเรียร์เพื่อสร้างเส้นฐานสำหรับการวิเคราะห์แนวโน้ม

3. การเตรียมข้อมูลสำหรับแบบจำลอง:
   - สร้างลำดับข้อมูลสำหรับการฝึกแบบจำลอง LSTM
   - ปรับขนาดข้อมูลให้อยู่ในช่วง 0-1 โดยใช้ MinMaxScaler
   - แบ่งข้อมูลเป็นชุดฝึกอบรม (80%) และชุดทดสอบ (20%)

4. การสร้างและฝึกแบบจำลอง:
   - สร้างแบบจำลอง LSTM แบบ bidirectional พร้อม dropout เพื่อป้องกัน overfitting
   - ใช้ early stopping และการปรับอัตราการเรียนรู้แบบลดลงเมื่อเกิดภาวะคงที่
   - ฝึกแบบจำลองด้วยข้อมูลชุดฝึกอบรม

5. การประเมินแบบจำลอง:
   - ทำนายค่าบนชุดทดสอบและคำนวณเมตริกประสิทธิภาพ (RMSE, MAE, R²)
   - สร้างกราฟแสดงความสัมพันธ์ระหว่างค่าจริงและค่าที่ทำนายได้

6. การพยากรณ์:
   - ใช้แบบจำลองที่ฝึกแล้วเพื่อพยากรณ์ระดับ CO สำหรับ 7 วันถัดไป
   - สร้างภาพรวมของการพยากรณ์พร้อมข้อมูลประวัติและเส้นฐานฟูเรียร์

7. การรายงานผล:
   - บันทึกผลลัพธ์ทั้งหมดเป็นไฟล์ภาพและไฟล์ข้อความ
   - สรุปสถิติเชิงพรรณนาและประสิทธิภาพของแบบจำลอง

เทคนิคพิเศษที่ใช้:
- การแปลงฟูเรียร์สำหรับการสร้างเส้นฐานแนวโน้ม
- LSTM แบบ bidirectional สำหรับการจับรูปแบบทั้งจากอดีตไป未來และ未来ไปอดีต
- Dropout และ early stopping เพื่อป้องกัน overfitting
- การปรับอัตราการเรียนรู้แบบลดลงเมื่อเกิดภาวะคงที่

การประยุกต์ใช้:
แบบจำลองนี้สามารถใช้สำหรับการพยากรณ์ระดับมลพิษทางอากาศในเขตเมือง 
และช่วยในการวางแผนจัดการคุณภาพอากาศในอนาคต
"""

# บันทึกคำอธิบายขั้นตอนการทำงาน
with open('procedure_description_th.txt', 'w', encoding='utf-8') as f:
    f.write(procedure_description)

print("คำอธิบายขั้นตอนการทำงานได้ถูกบันทึกไว้ในไฟล์: procedure_description_th.txt")
```
